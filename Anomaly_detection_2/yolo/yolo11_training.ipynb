## YOLO11-small segmentation training

```python
from ultralytics import YOLO

model = YOLO("/wrk/main/yolo/yolo_weights/yolo11s-seg.pt")

results = model.train(data="/wrk/main/yolo/data.yaml", epochs=100, device=0, seed=42)
```

```python
model.val(verbose=True, visualize=True)
```

## Classification metrics

```python
from pathlib import Path
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from ultralytics import YOLO

images_dir = Path("/wrk/main/yolo/yolo_data/images/val")

image_paths = [str(p) for p in images_dir.iterdir()]
true_labels = []
pred_labels = []

# Из-за особенностей директории с данными, можно так костыльно было заполнить true_labels 
for i in range(37):
    true_labels.append(1)

for i in range(37):
    true_labels.append(0)

model = YOLO("/wrk/main/yolo/runs/segment/train/weights/best.pt")

for img in image_paths:
    results = model(img)
    boxes = results[0].boxes
    if len(boxes) > 0:
        pred_labels.append(1)
    else:
        pred_labels.append(0)

acc = accuracy_score(true_labels, pred_labels)
prec = precision_score(true_labels, pred_labels)
rec = recall_score(true_labels, pred_labels)
f1 = f1_score(true_labels, pred_labels)

print(acc)
print(prec)
print(rec)
print(f1)
```
